## A Basic Pseudorandom Number Generator

### 1. Start with Seed
Seed is a starting state or number from which the sequence starts.

```python
X_0 =7
```

### 2. Use Linear Congruuential generator (LCG)
X_n+1 = (a*X_n + c ) * mod(m)

- X_n = Current State
- X_n+1 = next state
- a= multiplier
- c = increment
- m = modulus
- X_0 = seed

```python
#Intuition

def prng(a:int, c: int, m:int)->int:
    return (a*X_0 + c ) % m

# random numbers
a = 5
c= 3
m = 16

print(prng(a, c, m))
```

### As a Python Class

```python
class Randomizer:
    def __init__(self, seed, a, c, m):
        self.state = seed
        self.a = a
        self.c = c
        self.m = m

    def next(self):
        self.state = (a * self.state + 3) % 16
        return self.state
    
```

```python
r = Randomizer(7, a, c, m)

print(r.next())
print(r.next())
print(r.next())
```

```python
r1 = Randomizer(123, 1664525 , 1013904223, 2**32)

for i in range(0,5):
    print(r1.next())
```

### Observations
- Running it with the same number would always produce the same number
- Modulus helps in keeping the ranges within a limit. 


# Variables in Classes

```python
class Counter:
    class_counter= 10

    def __init__(self, counter):
        self.counter= counter

    def increment(self):
        self.counter = self.counter + 1
        Counter.class_counter = self.class_counter + 1

    def decrement(self):
        self.counter = self.counter - 1
        Counter.class_counter = self.class_counter - 1

    @staticmethod
    def counter_reset(new_counter:int):
        Counter.class_counter = new_counter
```

### Instance variables

```python
c1 = Counter(10)
c2 = Counter(10)

c1.increment()
c1.increment()
print(c1.counter)

c2.decrement()
print(c2.counter)
```

### Class variable

```python
c1 = Counter(10)
c2 = Counter(10)

c1.increment()
c1.increment()

print(c1.class_counter)
print(c2.class_counter)
print(Counter.class_counter)
```

### Static Code

```python
c1.counter_reset(13)


print(c1.class_counter)
print(c2.class_counter)
print(Counter.class_counter)

Counter.counter_reset(18)

print(c1.class_counter)
print(c2.class_counter)
print(Counter.class_counter)
```
