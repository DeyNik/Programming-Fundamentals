## Enums

````python
from enum import Enum, StrEnum

class Status(Enum):
    SUCCESS="success"

print(Status.SUCCESS.value)

class Status1(StrEnum):
    SUCCESS= "success"

print(Status1.SUCCESS)
````

## Bytes exists in Python - and since computers ultimately store/tramsmit data as bytes, this helps

````python
text="hello"
byte=b"hello"

print(type(text))
print(type(byte))
````

````python
data = text.encode("utf-8")
print(data)
print(data.decode("utf-8"))
print(type(data))
````

````python
with open("file.txt","r", encoding="utf-8") as f:
    text=f.read()

print(text)

with open("file.txt","rb") as f:
    text=f.read()

print(text)
````
